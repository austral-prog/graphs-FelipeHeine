## Line Plot

This graph shows how a quadratic function grows as x increases.
The curve illustrates a non-linear trend where values increase faster over time.
I chose a line plot because it is ideal for visualizing continuous data and mathematical functions.

```python
# Generate data
x = np.arange(0, 10)
y = x**2

# Create plot
plt.plot(x, y, color='blue', marker='o', linestyle='--', linewidth=2, label='y = x²')

# Labels and title
plt.title('Quadratic Growth Over Time')
plt.xlabel('X values (units)')
plt.ylabel('Y values (units)')

# Legend
plt.legend()

# Save image
plt.savefig('imgs/line_plot.png', dpi=300, bbox_inches='tight')

plt.show()
```

## Bar Chart

This bar chart compares sales across different product categories.
It clearly highlights which category performs best and worst.
I chose a bar chart because it is effective for comparing discrete groups.

```python
# Data
categories = ['A', 'B', 'C', 'D']
values = [10, 20, 15, 25]

# Create bar chart
plt.bar(categories, values, color='green', label='Sales')

# Labels and title
plt.title('Sales by Category')
plt.xlabel('Category')
plt.ylabel('Sales (units)')

# Legend
plt.legend()

# Save image
plt.savefig('imgs/bar_chart.png', dpi=300, bbox_inches='tight')

plt.show()
```

## Histogram

This histogram shows the distribution of randomly generated data.
It helps visualize how frequently values occur within certain ranges.
I chose a histogram because it is ideal for understanding data distribution.

```python
# Generate random data
data = np.random.randn(1000)

# Create histogram
plt.hist(data, bins=30, color='purple', edgecolor='black', label='Frequency')

# Labels and title
plt.title('Distribution of Random Data')
plt.xlabel('Value')
plt.ylabel('Frequency')

# Legend
plt.legend()

# Save image
plt.savefig('imgs/histogram.png', dpi=300, bbox_inches='tight')

plt.show()
```

## Scatter Plot

This scatter plot shows the relationship between two randomly generated variables.
It helps identify whether there is any visible correlation between them.
I chose a scatter plot because it is best for visualizing relationships between variables.

```python
# Generate data
x = np.random.rand(50)
y = np.random.rand(50)

# Create scatter plot
plt.scatter(x, y, color='red', marker='x', label='Data points')

# Labels and title
plt.title('Relationship Between Variables')
plt.xlabel('X values')
plt.ylabel('Y values')

# Legend
plt.legend()

# Save image
plt.savefig('imgs/scatter_plot.png', dpi=300, bbox_inches='tight')

plt.show()
```

## Box Plot

This box plot summarizes the distribution of two datasets.
It displays median, quartiles, and potential outliers in a compact form.
I chose a box plot because it is useful for comparing distributions statistically.


```python
# Generate data
data1 = np.random.randn(100)
data2 = np.random.randn(100) + 2

# Create box plot
plt.boxplot([data1, data2], tick_labels=['Dataset 1', 'Dataset 2'])

# Labels and title
plt.title('Comparison of Two Datasets')
plt.xlabel('Datasets')
plt.ylabel('Values')

# Save image
plt.savefig('imgs/custom_plot.png', dpi=300, bbox_inches='tight')

plt.show()
```
